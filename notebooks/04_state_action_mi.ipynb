# State-Action Mutual Information

Compute empirical discrete mutual information between state hashes and actions.

```python
from pathlib import Path
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path('../src').resolve()))
from mate.analysis.mi import mutual_information_by_group

sns.set_theme(style='whitegrid')
rollouts_path = Path('../runs/processed/rollouts.csv')
rollouts = pd.read_csv(rollouts_path) if rollouts_path.exists() else pd.DataFrame()
rollouts.head()
```

```python
mi = mutual_information_by_group(rollouts) if not rollouts.empty else pd.DataFrame()
mi.head()
```

```python
if not mi.empty:
    mi['label'] = mi['condition'].astype(str) + ' ' + mi['intervention'].astype(str)
    plt.figure(figsize=(9, 4))
    sns.barplot(data=mi, x='label', y='mi', errorbar='se')
    plt.title('State-action mutual information')
    plt.xlabel('Condition / intervention')
    plt.ylabel('MI')
    plt.xticks(rotation=25)
    plt.show()
```

# Intervention Effects

Compare C1 normal, marker deletion, and marker shuffle evaluation rollouts.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
rollouts_path = Path('../runs/processed/rollouts.csv')
rollouts = pd.read_csv(rollouts_path) if rollouts_path.exists() else pd.DataFrame()
rollouts.head()
```

```python
if not rollouts.empty:
    episodes = rollouts.groupby(['run_id', 'condition', 'seed', 'intervention', 'episode_index']).agg(
        total_return=('reward', 'sum'),
        steps=('t', 'max'),
        success=('done', 'max')
    ).reset_index()
    episodes['steps'] = episodes['steps'] + 1
    episodes.head()
```

```python
if not rollouts.empty:
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))
    sns.barplot(data=episodes, x='intervention', y='success', errorbar='se', ax=axes[0])
    axes[0].set_title('Success rate')
    sns.barplot(data=episodes, x='intervention', y='steps', errorbar='se', ax=axes[1])
    axes[1].set_title('Episode length')
    sns.barplot(data=episodes, x='intervention', y='total_return', errorbar='se', ax=axes[2])
    axes[2].set_title('Return')
    plt.tight_layout()
    plt.show()
```

```python
from pathlib import Path
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sys.path.insert(0, str(Path('../src').resolve()))
from mate.analysis.roles import classify_marker_roles, role_entropy, parse_json_list

sns.set_theme(style='whitegrid')
rollouts_path = Path('../runs/processed/rollouts.csv')
markers_path = Path('../runs/processed/markers.csv')
rollouts = pd.read_csv(rollouts_path) if rollouts_path.exists() else pd.DataFrame()
markers = pd.read_csv(markers_path) if markers_path.exists() else pd.DataFrame()
rollouts.head()
```

```python
roles = classify_marker_roles(rollouts, markers) if not rollouts.empty and not markers.empty else pd.DataFrame()
entropy = role_entropy(roles) if not roles.empty else pd.DataFrame()
roles.head()
```

```python
if not roles.empty:
    plt.figure(figsize=(10, 4))
    sns.barplot(data=roles, x='role', y='frequency', hue='intervention', errorbar='se')
    plt.title('Marker role frequencies')
    plt.xlabel('Role')
    plt.ylabel('Frequency')
    plt.xticks(rotation=25)
    plt.show()
```

```python
if not entropy.empty:
    plt.figure(figsize=(8, 4))
    sns.barplot(data=entropy, x='intervention', y='role_entropy', errorbar='se')
    plt.title('Marker role diversity')
    plt.xlabel('Intervention')
    plt.ylabel('Entropy')
    plt.show()
```

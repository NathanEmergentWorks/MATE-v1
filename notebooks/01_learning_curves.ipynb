# Learning Curves

Load `episodes.csv` and compare training return and success rate for C0 and C1.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
episodes_path = Path('../runs/processed/episodes.csv')
episodes = pd.read_csv(episodes_path) if episodes_path.exists() else pd.DataFrame()
episodes.head()
```

```python
if not episodes.empty:
    plt.figure(figsize=(10, 4))
    sns.lineplot(data=episodes, x='episode_index', y='total_return', hue='condition', errorbar='se')
    plt.title('Training return')
    plt.xlabel('Episode')
    plt.ylabel('Total return')
    plt.show()
```

```python
if not episodes.empty:
    success = episodes.sort_values(['run_id', 'episode_index']).copy()
    success['success_rate'] = success.groupby('run_id')['success'].transform(lambda s: s.rolling(50, min_periods=1).mean())
    plt.figure(figsize=(10, 4))
    sns.lineplot(data=success, x='episode_index', y='success_rate', hue='condition', errorbar='se')
    plt.title('Training success rate')
    plt.xlabel('Episode')
    plt.ylabel('Rolling success rate')
    plt.ylim(0, 1)
    plt.show()
```
